<h2 style="color:#1e4a76; margin-top:28px;">Justificación de las reglas</h2>

<div style="background:#f0f7ff; border-left:6px solid #2a7de1; padding:18px 22px; border-radius:0 12px 12px 0; box-shadow: 0 2px 8px rgba(42,125,225,0.08);">
  <p style="margin:0 0 10px 0; color:#1a3b5c;">
    Las reglas combinan el nivel de endeudamiento relativo
    (<code style="background:#eef4fc; color:#1a5ca8; padding:2px 6px; border-radius:4px;">monto / ingreso</code>)
    y la confiabilidad previa del solicitante. Para clientes <b>sin antecedentes de mora</b>,
    se otorgan márgenes más amplios de crédito antes de rechazar o condicionar.
  </p>
  <p style="margin:0; color:#1a3b5c;">
    Para clientes <b>con historial moroso</b>, la tolerancia al endeudamiento es estrictamente
    menor para resguardar la liquidez del banco frente al riesgo de impago.
  </p>
</div>

</div>

<div style="font-family: 'Segoe UI', Roboto, Arial, sans-serif; background: linear-gradient(135deg,#e8f0fe 0%,#f3f7fb 100%); padding: 20px 24px; border-radius: 16px; color:#1a2b3c;">

<h2 style="color:#1e4a76; margin:0; font-weight:600;">
Implementación del código
</h2>

</div>

```python
def agente_credito(ingreso_mensual, monto_solicitado, tiene_historial_moroso):
    if ingreso_mensual <= 0:
        return "rechazar", "El ingreso mensual debe ser mayor a cero."
    
    relacion = monto_solicitado / ingreso_mensual
    
    if tiene_historial_moroso:
        if relacion <= 2.0:
            return "aprobar con condiciones", f"Historial moroso registrado, pero relación monto/ingreso baja ({relacion:.2f} <= 2.0)."
        else:
            return "rechazar", f"Historial moroso registrado y relación monto/ingreso alta ({relacion:.2f} > 2.0)."
    else:
        if relacion <= 4.0:
            return "aprobar", f"Historial limpio y relación monto/ingreso óptima ({relacion:.2f} <= 4.0)."
        elif relacion <= 7.0:
            return "aprobar con condiciones", f"Historial limpio con relación monto/ingreso moderada ({relacion:.2f} entre 4.0 y 7.0)."
        else:
            return "rechazar", f"Relación monto/ingreso excesiva ({relacion:.2f} > 7.0)."
```

<h2 style="color:#1e4a76; margin:0; font-weight:600;">
Simulación y pruebas
</h2>

</div>

```python
pruebas = [
    (5000, 10000, False), # Relación 2.0 (Limpio -> Aprobar)
    (5000, 20000, False), # Relación 4.0 (Borde Limpio -> Aprobar)
    (5000, 25000, False), # Relación 5.0 (Limpio -> Aprobar con condiciones)
    (5000, 40000, False), # Relación 8.0 (Limpio -> Rechazar)
    (3000, 6000, True),   # Relación 2.0 (Borde Moroso -> Aprobar con condiciones)
    (3000, 9000, True)    # Relación 3.0 (Moroso -> Rechazar)
]

print("--- PRUEBAS DE EVALUACIÓN DE CRÉDITO ---")
for ingreso, monto, moroso in pruebas:
    accion, motivo = agente_credito(ingreso, monto, moroso)
    print(f"Ingreso: S/{ingreso} | Monto: S/{monto} | Moroso: {moroso} => {accion.upper()}: {motivo}")
```

<h2 style="color:#1e4a76; margin:0; font-weight:600;">
Visualización
</h2>

</div>

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
num_muestras = 200

# Generar datos 
ingresos = np.random.uniform(1500, 10000, num_muestras)
montos = np.random.uniform(2000, 60000, num_muestras)
historiales = np.random.choice([True, False], size=num_muestras, p=[0.3, 0.7])

acciones = []
relaciones = []

for i in range(num_muestras):
    acc, _ = agente_credito(ingresos[i], montos[i], historiales[i])
    acciones.append(acc)
    relaciones.append(montos[i] / ingresos[i])

#Colores
colores_map = {
    "aprobar": "green",
    "aprobar con condiciones": "orange",
    "rechazar": "red"
}
colores = [colores_map[acc] for acc in acciones]

plt.figure(figsize=(10, 6))
for acc in colores_map:
    idx = [j for j, a in enumerate(acciones) if a == acc]
    plt.scatter(
        [relaciones[j] for j in idx], 
        [ingresos[j] for j in idx], 
        c=colores_map[acc], 
        label=acc.capitalize(), 
        alpha=0.7, 
        edgecolors='k'
    )

plt.title("Evaluación de Solicitudes de Crédito (Agente Inteligente)")
plt.xlabel("Relación Monto Solicitado / Ingreso Mensual")
plt.ylabel("Ingreso Mensual (Soles)")
plt.axvline(x=2.0, color='gray', linestyle='--', label='Límite Moroso (2.0)')
plt.axvline(x=4.0, color='blue', linestyle='--', label='Límite Optimo (4.0)')
plt.axvline(x=7.0, color='purple', linestyle='--', label='Límite Máximo (7.0)')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.tight_layout()
plt.show()
```
