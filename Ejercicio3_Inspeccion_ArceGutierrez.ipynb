<h2 style="color:#1e4a76; margin-top:28px;">Justificación de las reglas</h2>

<div style="background:#f0f7ff; border-left:6px solid #2a7de1; padding:18px 22px; border-radius:0 12px 12px 0; box-shadow: 0 2px 8px rgba(42,125,225,0.08);">
  <p style="margin:0 0 10px 0; color:#1a3b5c;">
    Se requiere <b>perfección absoluta</b> (0 defectos) para el despacho directo automatizado
    de productos. Un recuento de <b>1 a 3 píxeles defectuosos</b> se deriva a revisión manual
    para evaluar si el fallo es corregible o cosmético, reduciendo el desperdicio.
  </p>
  <p style="margin:0; color:#1a3b5c;">
    Si el producto supera los <b>3 píxeles defectuosos</b>, se descarta directamente por ser
    irrecuperable y evitar pérdidas de tiempo de reproceso.
  </p>

```python
import numpy as np

def agente_inspeccion(imagen):
    """
    Percepción (S): matriz 5x5 de NumPy (0 = normal, 1 = defectuoso)
    """
    defectos = int(np.sum(imagen == 1))
    
    if defectos == 0:
        return "aprobar", f"Producto libre de imprecisiones ({defectos} defectos detectados)."
    elif defectos <= 3:
        return "revisión manual", f"Defectos menores detectados ({defectos} defectos). Requiere inspección visual."
    else:
        return "rechazar", f"Producto fuera de tolerancia ({defectos} defectos detectados)."
```
